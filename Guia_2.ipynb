```python
# Un usuario decide inverti $500 a principio de cada mes. La cuenta de inversion ofrece un
# rendimiento anual del 6%(que equivale a un 0.5% mensual). Desarrollar un program
# que utilice un bucle for para calcular y mostrar mes a mes como crece su cuenta durante
# un año(12 meses), aplicando el interes compuesto sobre el saldo acumulado. Al finalizar,
# programa debe imprimir el saldo total final y la ganancia obtenida solo por intereses.
```

```python
# Definicion de parametros
deposito_mensual = 500
tasa_mensual = 0.005
meses = 12
```

```python
# inicializacion acumuladores
saldo_acumulado = 0.0
total_depositado = 0.0
```

```python
# Bucle
for mes in range(1, meses + 1):
  saldo_inicial_mes = saldo_acumulado # Saldo antes del depósito
  saldo_acumulado = saldo_acumulado + deposito_mensual # Se añade el depósito de este mes
  interes_mes = saldo_acumulado * tasa_mensual # Se calcula el interés sobre el nuevo saldo
  saldo_acumulado = saldo_acumulado + interes_mes # Se añade el interés al saldo
  total_depositado = total_depositado + deposito_mensual # Acumulamos el total depositado
  print(f"{"Mes":<6} | {"Depósito":<11} | {"Saldo Pre-Int":<14} | {"Interés Ganado":<15} | {"Saldo Final":<12}")
  print(f"{mes:^6d} | ${deposito_mensual:,.2f}  | ${saldo_inicial_mes:,.2f}   | ${interes_mes:,.2f}      | ${saldo_acumulado:,.2f}")
```

```python
ganancia_intereses = saldo_acumulado - total_depositado
print(f"• Total de capital aportado:         ${total_depositado:,.2f} USD")
print(f"• Total ganado por intereses (NETO):  ${ganancia_intereses:,.2f} USD")
print(f"• SALDO ACUMULADO FINAL EN CUENTA:   ${saldo_acumulado:,.2f} USD")
```

```python
# EJERCICIO 10
# El promedio movil simple(SMA) es una manera de suavizar las fluctuaciones de precios.
# Dada una serie de precios historicos: historico = [45.5, 46.2, 46.8. 45.9, 47.1. 47.5]
# Haga una nueva lista con el SMA de 3 dias. Notar que la nueva lista tiene menos elementos
# que la original
```

```python
# Serie histrica de precios
historico = [45.5, 46.2, 46.8, 45.9, 47.1, 47.5, 48.0, 47.8]
# Inicializacion de la lista de promedios moviles vacia en la RAM
sma_3 = []
# Tamaño de la ventana
K = 3
```

```python
# Bucle indexado que arranca en K-1(indice 2) para tener ventana
# completa. Range(2,8) recorrera los indices: 2, 3, 4, 5, 6, 7
for i in range(K -1, len(historico)):
  p_hoy = historico[i]
  p_ayer = historico[i - 1]
  p_anteayer = historico[i - 2]
  promedio = (p_hoy + p_ayer + p_anteayer)/ 3
  sma_3.append(promedio) # Añadir el promedio a la lista
```

```python
print(f"Lista original de precios: {historico}")
print(f"SMA de {K} días: {sma_3}")
```

```python
# Ejercicio 15
# Evaluar si un proyecto de inversion es rentable calculando su VPN
# Definir directamente la inversion inicial con un valor negativo(por ejemplo, -10000) y
# guardar los flujos de fondos esperados para los proximos 4 años en una lista, como flujos =
# [3000, 4000, 5000, 2000]. Luego ingresar mediante un input() que ingrese la tasa de descuento
# esperada en decimales(ej: 0.10)
# 1. Inicializar un acumulador que empiece en 0(para ir guardando la suma de los resultados) y un contador de tiempo(t)
# inicializado en 1 para representar el primer año.
# 2. Hacer un bucle que, tome el flujo, correspondiente a ese año, y calcular su valor presente dividiendolo por
# (1+r) y suma ese resultado a tu acumulador
# 3. Al salir del bucle, sumar la inversion inicial a ese total acumulador para obtener el
# VPN definitivo. Imprimir el resultado final redondeado a dos decimales. y usando un
# condicional definir si el VPN es mayor a cero, muestra el mensaje "Proyecto viable"; de lo contrario,
# indica "Proyecto No viable"
```

```python
inversion_inicial = -10000
flujos = [3000, 4000, 5000, 2000]
tasa = float(input('Ingrese la tasa de descuento en decimales'))
```

```python
acumulador_flujo = 0
t = 1
```

```python
for flujo in flujos:
  valor_presente = flujo / (1 + tasa)** t
  acumulador_flujo = acumulador_flujo + valor_presente
  t = t + 1
```

```python
vpn = inversion_inicial + acumulador_flujo

print(f"El VPN del proyecto es: ${vpn:.2f}")

if vpn > 0:
  print("Proyecto viable")
else:
  print("Proyecto No viable")
```

```python
# EJERCICIO 16
# Un proyecto de inverion requiere $100 hoy y devuelve $120 en un año. Desarrollar un
# programa que utilice un bucle while para encontrar la TIR por busqueda
# incremental. El algoritmo debe comenzar con una tasa r = 0 e ir
# incrementandola de a 0.0001 hasta el valor presente neto (VPN= -100 + 120/(1+r))
# sea aproximadamente 0(menor a 0.01). Imprimir la tasa resultante en forma porcentual
```

```python
# 1. Datos iniciales del proyecto de inversión
inversion = -100
retorno = 120

# 2. Variables de control para la búsqueda incremental
r = 0.0            # Tasa de partida (0.0%)
paso = 0.0001      # Resolución del ajuste (0.01%)

# 3. Calculamos el primer VPN de prueba en la RAM
vpn = inversion + retorno / (1 + r)

# 4. Bucle while: se ejecuta mientras el VPN no sea aproximadamente cero
# Usamos ">= 0.01" como condición porque el VPN va cayendo a medida que la tasa sube.
while vpn >= 0.01:
    r = r + paso                       # Incrementamos la tasa de descuento
    vpn = inversion + retorno / (1 + r) # Recalculamos el VPN con la nueva tasa

# 5. Mostramos el resultado final formateado

# Multiplicamos r por 100 para mostrarla como porcentaje
print("TIR aproximada encontrada:", round(r * 100, 2), "%")
print("VPN final alcanzado:      $", round(vpn, 4))
```

```python
# EJERCICIO 17 Diseñar un algoritmo que calcule la TIR de un proyecto mediante el método de tanteo
# (búsqueda incremental simple). El proyecto tiene una inversión inicial de -1000 y flujos
# futuros representados en una lista: [400, 400, 400].
# Definir las variables de partida: (i) tasa
# _prueba se inicia en 0.0. Esta es la tasa de
#descuento a ajustar (ii) paso se establece en 0.001. Define la resolución del algoritmo,
# indicando cuánto se incrementará la tasa en cada intento (iii) vpn_prueba se inicializa
# con un valor positivo arbitrario (por ejemplo, 100) simplemente para asegurar que el
# algoritmo ingrese al ciclo de evaluación (iv) iteraciones es un contador que arranca en 0.
# El núcleo del programa es un bucle que se repetirá mientras el vpn
# _prueba se mantenga mayor a 0. Dentro de este ciclo, en cada iteración se debe:
# 1. Calcular el VPN actual, descontando los flujos del proyecto con la tasa_prueba
# vigente y sumando la inversión inicial.
# 2. Actualizar la variable vpn_prueba con el resultado de este nuevo cálculo.
# 3. Si el VPN sigue arrojando un resultado positivo, significa que la tasa de descuento
# aún es baja. Por lo tanto, se incrementa la tasa_prueba sumándole el valor del
# paso y se suma 1 al contador de iteraciones.
# Dado que el VPN disminuye a medida que la tasa aumenta, en algún momento el
# resultado se volverá negativo. Cuando esto ocurra, se habrá encontrado la aproximación
# a la TIR y el ciclo se detendrá automáticamente. Al finalizar, el programa se debe imprimir la tasa_prueba final formateada como un porcentaje con dos decimales, junto
# con la cantidad total de iteraciones que fueron necesarias para alcanzar el resultado.
```

```python
# 1. Datos iniciales del proyecto
inversion_inicial_tir = -1000
flujos_tir = [400, 400, 400]

# 2. Variables de partida
tasa_prueba = 0.0          # Tasa de descuento inicial
paso = 0.001             # Incremento de la tasa en cada iteración
vpn_prueba = 100         # Valor positivo arbitrario para asegurar la entrada al bucle
iteraciones = 0          # Contador de iteraciones

# 3. Bucle para encontrar la TIR por búsqueda incremental
while vpn_prueba > 0:
    vpn_actual = inversion_inicial_tir
    for i, flujo in enumerate(flujos_tir):
        vpn_actual += flujo / ((1 + tasa_prueba) ** (i + 1))

    vpn_prueba = vpn_actual # Actualizar vpn_prueba con el VPN calculado

    if vpn_prueba > 0: # Si el VPN sigue siendo positivo, la tasa es muy baja
        tasa_prueba += paso  # Incrementar la tasa
        iteraciones += 1     # Incrementar el contador de iteraciones

# 4. Imprimir los resultados finales
print(f"Tasa Interna de Retorno (TIR) aproximada: {tasa_prueba * 100:.2f}%")
print(f"Cantidad total de iteraciones: {iteraciones}")
```
