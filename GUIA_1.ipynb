```python
# Ejercicio 12
# Crear un programa que calcule la comisión de un broker basándose en el volumen
# operado mensualmente. El script debe solicitar el monto de la operación mediante input()
# y convertirlo a float. La estructura del algoritmo debe resolver el cálculo de dos formas
# distintas dentro del mismo script:
# 1. Mediante Lógica Anidada: Utilizar estructuras if/else una dentro de otra para
# segmentar los tramos de comisión (Hasta 10.000: 0,5%; de 10.001 a 50.000:
# 0,3%; más de 50.000: fija de 100 + 0,1%).
# 2. Mediante Operadores Lógicos: Replicar el cálculo utilizando una estructura plana
# con elif y el operador and para definir los rangos de manera explícita.
# El programa debe imprimir el resultado de ambos métodos, asegurándose de que
# coincidan, y mostrar la comisión final redondeada a 2 decimales.

```

```python
# Entrada de datos
monto = float(input("Ingrese el monto de la operacion: "))
```

```python
# METODO 1
if monto <= 10000:
    comision_anidada = monto * 0.005
else:
    if monto <= 50000:
        comision_anidada = monto * 0.003
    else:
        comision_anidada = 100 + (monto * 0.001)
```

```python
# --- MÉTODO 2: Estructura Plana (Operadores Lógicos) ---
if monto <= 10000:
    comision_plana = monto * 0.005
elif monto > 10000 and monto <= 50000:
    comision_plana = monto * 0.003
elif monto > 50000:
    comision_plana = 100 + (monto * 0.001)
```

```python
coinciden = round(comision_anidada, 2) == round(comision_plana, 2)
```

```python
# 2. Emisión del Reporte de Auditoría en Pantalla
print("=" * 55)
print("             AUDITORÍA DE COMISIONES - BROKER")
print("=" * 55)
print(f"• Monto de la Operación:         ${monto:,.2f} USD")
print("-" * 55)
print(f"• Comisión (Lógica Anidada):     ${comision_anidada:,.2f} USD")
print(f"• Comisión (Operadores Lógicos):  ${comision_plana:,.2f} USD")
print(f"• ¿Ambos cálculos coinciden?:     {coinciden} [✔]")
print("-" * 55)
print(f"• LIQUIDACIÓN FINAL DE COMISIÓN:  ${comision_plana:,.2f} USD")
print("=" * 55)
```
